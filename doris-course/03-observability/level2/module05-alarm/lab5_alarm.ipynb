## Alarm

This section explains how to configure alarms in Grafana while using DOG so that abnormal observability data can be detected in time.

Consider an example in which Doris metric data has already been collected. To monitor the cluster in real time, an alert should be triggered when `system.cpu.utilization` exceeds 98%. How can this alert be configured in Grafana?

First, sign in with your own Grafana account and password. If no credentials have been configured, the default values are `admin/admin`.

Select Alert rules:

<img src="images/alter_rule.png" alt="Opening Alert rules in Grafana" style="max-width:100%;height:auto;">

Next, select the expression. Select the `otel_metrics_gauge` table, use `value` as the query column, and set the filter condition to `metric_name = 'system.cpu.utilization'`:

<img src="images/config_grafana_1.png" alt="Configuring the Grafana alert query" style="max-width:100%;height:auto;">

Then configure the alert threshold:

<img src="images/config_threshold.png" alt="Configuring the Grafana alert threshold" style="max-width:100%;height:auto;">

Select a folder and group. If none exist, create them to save the alert information:

<img src="images/folder.png" alt="Selecting a folder and group for the Grafana alert" style="max-width:100%;height:auto;">

Select a notification method, such as email. If no notification method exists, create one:

<img src="images/email.png" alt="Selecting email as the Grafana notification method" style="max-width:100%;height:auto;">

<img src="images/contact_point.png" alt="Creating a Grafana contact point" style="max-width:100%;height:auto;">

Enter the notification details:

<img src="images/message.png" alt="Entering Grafana notification message details" style="max-width:100%;height:auto;">

After the configuration is complete, Grafana sends an email notification when `system.cpu.utilization` in `otel_metrics_gauge` reaches 98%.
